```python
#-- Step 1: Import Python Library. --#
import pandas as pd
#-- Step 2: Load the .csv File. --#
data = pd.read_csv('Titanic_Dataset.csv')
#-- Step 3: Display all data. --#
data
```

```python
#-- Import Matplotlib Library. --#
import matplotlib.pyplot as plt
#-- Step 1: Group the dataset by Passenger Class and Survived status, then pivot the table for plotting. --#
survival_counts = data.groupby(['Pclass', 'Survived']).size().unstack()
#-- Step 2: Create a bar chart using Pandas built-in plotting functionality. --#
ax = survival_counts.plot(kind='bar', figsize=(8, 5), color=['#ff9999', '#66b3ff'])
#-- Step 3: Set chart title and axis labels. --#
plt.title('Figure 10: Survival Count by Passenger Class', fontsize=14, fontweight='bold')
plt.xlabel('Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)', fontsize=12)
plt.ylabel('Passenger Count', fontsize=12)
#-- Step 4: Customize the legend, ticks, and background grid for better readability. --#
plt.legend(['No (0)', 'Yes (1)'], title='Survived')
plt.xticks(ticks=[0, 1, 2], labels=['1st Class', '2nd Class', '3rd Class'], rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
#-- Step 5: Display the generated graph. --#
plt.tight_layout()
plt.show()
```
